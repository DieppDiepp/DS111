```python
def count_real_missing_robust(df):
    # Tạo bản sao để không ảnh hưởng data gốc
    df_temp = df.copy()

    # 1. Danh sách các "chuỗi" cần coi là Missing
    # Bao gồm chuỗi rỗng, chữ "None", "NaN", "nan", "null" (đề phòng LLM viết thường/hoa)
    fake_missing_values = ["", "None", "NaN", "nan", "Null", "null"]

    # Thay thế toàn bộ các chuỗi này trong dataframe thành np.nan thật sự
    # Dùng regex=False để tìm chính xác chuỗi
    df_temp.replace(fake_missing_values, np.nan, inplace=True)

    # 2. Xử lý riêng cho các cột dạng List (Roles, Tech_Stack...)
    # Vì replace ở trên không tác động được vào bên trong cái list [""], ["None"]
    list_cols = ['Roles', 'Province', 'Company_Location', 'Language', 
                 'Tech_Stack', 'Programming_Language', 'Skills']
    
    for col in list_cols:
        if col in df_temp.columns:
            # Logic: 
            # - Nếu là list rỗng [] -> NaN
            # - Nếu là list chứa chuỗi rỗng [""] -> NaN
            # - Nếu là list chứa chuỗi "None" ["None"] -> NaN
            def clean_list(x):
                if not isinstance(x, list): 
                    return x # Nếu đã là NaN do bước 1 thì bỏ qua
                
                # Nếu list rỗng
                if len(x) == 0: 
                    return np.nan
                
                # Nếu phần tử đầu tiên nằm trong danh sách "missing trá hình"
                if len(x) == 1 and str(x[0]) in fake_missing_values:
                    return np.nan
                
                return x

            df_temp[col] = df_temp[col].apply(clean_list)

    # 3. Đếm lại bằng isnull (lúc này đã sạch)
    return df_temp.isnull().sum()

# --- CHẠY THỬ ---
print("Số lượng missing thực tế (bao gồm cả string 'None', 'NaN'):")
print(count_real_missing_robust(data_API))
```

```python
import ast
def safe_convert_to_list(x):
    if isinstance(x, list):
        return x  # Nếu đã là list thì giữ nguyên
    if pd.isna(x) or x == "" or x == "[]":
        return [] # Trả về list rỗng nếu không có dữ liệu
    try:
        # ast.literal_eval giúp chuyển string thành cấu trúc python an toàn
        return ast.literal_eval(str(x)) 
    except (ValueError, SyntaxError):
        return [] # Nếu lỗi định dạng thì trả về rỗng

# 1. Áp dụng chuyển đổi cho các cột cần thiết
cols_to_fix = ['Tech_Stack', 'Programming_Language', 'Province', 'Language', 'Roles', 'Company_Location', 'Skills']

for col in cols_to_fix:
    if col in data_API.columns:
        # Chuyển đổi dữ liệu sang List thật sự
        data_API[col] = data_API[col].apply(safe_convert_to_list)

# 2. Sau khi đã convert xong, chạy lại đoạn code thống kê cũ
flattened_Tech = list(itertools.chain.from_iterable(data_API['Tech_Stack']))
flattened_Programming = list(itertools.chain.from_iterable(data_API['Programming_Language']))
flattened_Province = list(itertools.chain.from_iterable(data_API['Province']))
flattened_Language = list(itertools.chain.from_iterable(data_API['Language']))
flattened_Roles = list(itertools.chain.from_iterable(data_API['Roles']))
flattened_Company_Location = list(itertools.chain.from_iterable(data_API['Company_Location']))
flattened_Skills = list(itertools.chain.from_iterable(data_API['Skills']))

# 3. Tạo bảng đếm
tech_counts = pd.Series(flattened_Tech).value_counts().reset_index()
tech_counts.columns = ['Tech_Stack', 'Count']

programming_counts = pd.Series(flattened_Programming).value_counts().reset_index()
programming_counts.columns = ['Programming_Language', 'Count']

province_counts = pd.Series(flattened_Province).value_counts().reset_index()
province_counts.columns = ['Province', 'Count']

language_counts = pd.Series(flattened_Language).value_counts().reset_index()
language_counts.columns = ['Language', 'Count']

roles_counts = pd.Series(flattened_Roles).value_counts().reset_index()
roles_counts.columns = ['Roles', 'Count']

company_location_counts = pd.Series(flattened_Company_Location).value_counts().reset_index()
company_location_counts.columns = ['Company_Location', 'Count']

skills_counts = pd.Series(flattened_Skills).value_counts().reset_index()
skills_counts.columns = ['Skills', 'Count']

# 4. Hiển thị kết quả đúng (toàn bộ)
num_rows_to_display = 50
pd.set_option('display.max_rows', num_rows_to_display)
print("Tech Stack Counts:")
tech_counts.head(num_rows_to_display)
```

```python
# 1. Định nghĩa các từ khóa "thừa" thường gặp trong địa chỉ
keywords = ["THÀNH PHỐ", "QUẬN", "XÃ", "HUYỆN", "TỈNH", "XÓM", "ĐƯỜNG", "HẺM", "TP ", "TP.", "Q.", "PHƯỜNG"]

# 2. Hàm kiểm tra (Check)
def has_admin_keyword(data_list):
    # Nếu không phải list hoặc rỗng -> Bỏ qua
    if not isinstance(data_list, list) or len(data_list) == 0:
        return False
    
    # Duyệt từng địa điểm trong list
    for item in data_list:
        if not isinstance(item, str): continue
            
        item_upper = item.upper()
        # Kiểm tra xem có từ khóa nào dính trong chuỗi không
        for kw in keywords:
            if kw in item_upper:
                return True
    return False

# 3. Áp dụng bộ lọc
print("Đang quét dữ liệu...")
mask_dirty_location = data_API['Province'].apply(has_admin_keyword) | \
                      data_API['Company_Location'].apply(has_admin_keyword)

# 4. Hiển thị kết quả
dirty_rows = data_API[mask_dirty_location]
dirty_rows[['Province', 'Company_Location']]
```

```python
print(f"Phát hiện {len(dirty_rows)} dòng chứa từ khóa hành chính (Tỉnh/Huyện/Quận...).")
replace_dict = {
    "THÀNH PHỐ HỒ CHÍ MINH": "HỒ CHÍ MINH",
    "BÌNH THẠNH - TP HCM": "HỒ CHÍ MINH",
    "QUẬN 3 - TP HCM": "HỒ CHÍ MINH",
    "PHÚ NHUẬN - TP HCM": "HỒ CHÍ MINH",
    "THỦ ĐỨC - TP HCM": "HỒ CHÍ MINH",
    "CẦU GIẤY - HÀ NỘI": "HÀ NỘI",
    "THẠCH THẤT - HÀ NỘI": "HÀ NỘI",
    "ĐỐNG ĐA - HÀ NỘI": "HÀ NỘI",
}

# 2. HÀM ÁP DỤNG MAPPING
def apply_mapping_strict(loc_list):
    # Nếu không phải list thì trả về nguyên trạng
    if not isinstance(loc_list, list):
        return loc_list
    
    cleaned_list = []
    for item in loc_list:
        if not isinstance(item, str):
            continue
            
        # Xóa khoảng trắng thừa 2 đầu để khớp chính xác với key trong từ điển
        item_key = item.strip()
        
        # LOGIC CHÍNH:
        # Nếu item_key có trong từ điển -> Lấy giá trị mới
        # Nếu không có -> Giữ nguyên giá trị cũ
        new_val = replace_dict.get(item_key, item_key)
        
        cleaned_list.append(new_val)
    
    # Loại bỏ trùng lặp và sắp xếp lại
    return sorted(list(set(cleaned_list)))

# 3. THỰC THI TRÊN 2 CỘT
cols_target = ["Province", "Company_Location"]

for col in cols_target:
    if col in data_API.columns:
        data_API[col] = data_API[col].apply(apply_mapping_strict)
```

Programming_Language trong Tech_Stack

```python
# Đếm số lượng Programming_Language lẫn lộn trong Tech_Stack
cntPL = 0
for tech in data_API['Tech_Stack'].explode().unique():
    if tech in OffProLang:
        print(f"Ngôn ngữ lập trình lẫn lộn: {tech}")
        cntPL += 1
print(f"Số lượng ngôn ngữ lập trình lẫn lộn trong Tech_Stack trước khi chuẩn hóa: {cntPL}")
```

```python
# Xử lí các ngôn ngữ lập trình lẫn lộn trong Tech_Stack
for idx, row in data_API.iterrows():
    tech_stack = row['Tech_Stack']
    prog_lang = row['Programming_Language']

    # Bỏ qua nếu tech_stack và prog_lang là null
    if tech_stack is None and prog_lang is None:
        continue

    # Chuyển về list để thao tác
    tech_stack_list = tech_stack if isinstance(tech_stack, list) else []
    prog_lang_list = prog_lang if isinstance(prog_lang, list) else []

    # Tìm các ngôn ngữ lập trình trong tech_stack
    langs_in_tech = [lang for lang in tech_stack_list if lang in OffProLang]

    # Xóa tất cả ngôn ngữ lập trình khỏi tech_stack
    tech_stack_list = [tech for tech in tech_stack_list if tech not in OffProLang]

    # Bổ sung vào prog_lang nếu chưa có
    for lang in langs_in_tech:
        if lang not in prog_lang_list:
            prog_lang_list.append(lang)

    # Gán lại giá trị cho DataFrame
    #Kiểm tra nếu tech_stack_list rỗng thì gán giá trị None
    if not tech_stack_list:
        tech_stack_list = None
    # Đảm bảo giá trị không trùng bằng set, nếu không phải rỗng
    data_API.at[idx, 'Tech_Stack'] = list(set(tech_stack_list)) if tech_stack_list else None
    data_API.at[idx, 'Programming_Language'] = list(set(prog_lang_list)) if prog_lang_list else None
```

```python
def normalize_field_smart(x):
    # Trường hợp 1: Giá trị là List (VD: Tech_Stack) -> Sắp xếp và nối lại
    # Ví dụ: ['java', ' python'] -> 'JAVA, PYTHON'
    #        [' python, 'java'] -> 'JAVA, PYTHON'    
    if isinstance(x, list):
        items = sorted(set([str(i).strip().upper() for i in x if i and str(i).strip() != '']))
        print(f"Normalized list from {x} to {items}")
        return ', '.join(items)
    
    # Trường hợp 2: Giá trị là String (VD: Company_Name) -> Chỉ viết hoa, KHÔNG ĐƯỢC XÓA
    elif isinstance(x, str):
        return x.strip().upper()
    
    # Trường hợp 3: Khác (VD: số năm kinh nghiệm) -> Ép kiểu string
    else:
        return str(x).strip().upper() if pd.notna(x) else ''

cols_dup = ['Old_Job_Title', 'Company_Name','Experience_Years', 'Job_Level', 'Academic_Level', 
            'Tech_Stack', 'Programming_Language', 'Province', 'Language', 'Roles', 'Company_Location']

# Chuẩn hóa các cột dạng list sang string
for col in cols_dup:
    check_unique_before_drop_dup[col] = check_unique_before_drop_dup[col].apply(normalize_field_smart)

# Loại bỏ trùng lặp, chỉ giữ lại dòng đầu tiên
check_unique_after_drop_dup = check_unique_before_drop_dup.drop_duplicates(subset=cols_dup, keep='first').reset_index(drop=True)
```

```python
# Sau khi đã loại trùng xong dưới dạng string, ta sẽ khôi phục lại các cột dạng list về đúng định dạng ban đầu để phục vụ cho việc chuẩn hóa về sau
def restore_string_to_list_clean(x):
    """Khôi phục từ String về List và xử lý UNKNOWN cùng lúc
    """
    # Nếu là NaN hoặc chuỗi rỗng -> Trả về ['UNKNOWN']
    if pd.isna(x) or (isinstance(x, str) and x.strip() == ''):
        return ['UNKNOWN']
    
    # Nếu là chuỗi -> Tách và XÓA KHOẢNG TRẮNG thừa (Quan trọng)
    if isinstance(x, str):
        return [item.strip() for item in x.split(',')]
    
    return [str(x)]

# QUY TRÌNH XỬ LÝ (PIPELINE)

#Chuẩn hóa dữ liệu sang String để chuẩn bị lọc
for col in cols_dup:
    if col in data_API.columns:
        data_API[col] = data_API[col].apply(normalize_field_smart)

num_samples_before = data_API.shape[0]

# Lọc trùng lặp (Drop Duplicates)
data_API = data_API.drop_duplicates(subset=cols_dup, keep='first').reset_index(drop=True)

# Khôi phục các cột List về định dạng danh sách chuẩn
for col in cols_dup:
    if col in data_API.columns:
        data_API[col] = data_API[col].apply(restore_string_to_list_clean)
```

```python
# BƯỚC 1: PHÂN RÃ DỮ LIỆU
# Explode: Nổ tung danh sách Roles thành các dòng riêng lẻ
# VD: ['Java', 'Backend'] → 2 dòng (Java), (Backend)
# Mục đích: Để tính median lương cho TỪNG role riêng biệt
df_exploded = data_API[['Roles', 'Salary_Min', 'Salary_Max']].explode('Roles')

# BƯỚC 2: TÍNH MEDIAN LƯƠNG CHO MỖI ROLE
# dropna: Loại bỏ những dòng không có dữ liệu lương (chỉ tính trên dữ liệu có sẵn)
# groupby: Nhóm theo Role
# median: Tính giá trị giữa (trung vị) của lương cho mỗi role
# Kết quả: Dictionary {'Java': 50M, 'Backend': 45M, ...}
median_min_by_role = df_exploded.dropna(subset=['Salary_Min']).groupby('Roles')['Salary_Min'].median() # Tính median lương tối thiểu
median_max_by_role = df_exploded.dropna(subset=['Salary_Max']).groupby('Roles')['Salary_Max'].median() # Tính median lương tối đa

# BƯỚC 3: ĐỊNH NGHĨA HÀM ƯỚC TÍNH LƯƠNG
def estimate_salary_from_list(roles_list, median_dict):
    """
    Input:  roles_list = ['Java', 'Backend'] (các role của 1 công việc)
            median_dict = {'Java': 50M, 'Backend': 45M, ...}
    Output: Lương ước tính (trung bình cộng của các role)
    """
    
    # KIỂM TRA: Nếu không phải list hoặc list rỗng → Trả NaN
    if not isinstance(roles_list, list) or len(roles_list) == 0:
        return np.nan  # NaN sẽ dễ xử lý khi fillna()
    
    # TÌM MEDIAN: Lấy median lương của từng role trong danh sách trên 1 công việc/tin tuyển dụng
    # VD: roles_list = ['Java', 'Backend'] 
    #     → vals = [50M, 45M]
    vals = []
    for r in roles_list:
        if r in median_dict:  # Chỉ lấy role có dữ liệu
            vals.append(median_dict[r])
    
    # TÍNH TRUNG BÌNH: Nếu tìm thấy ít nhất 1 role có dữ liệu
    # → Tính mean (trung bình cộng) của các median đó
    # VD: vals = [50M, 45M] → mean = (50+45)/2 = 47.5M
    # Nếu không tính trung bình thì sẽ có hiện tượng 1 công việc/tin tuyển dụng có nhiều mức lương khác nhau -> Chỉ cần 1 mức lương ước tính duy nhất
    if vals:
        return np.mean(vals)
    
    # NẾU KHÔNG TÌM THẤY: Không role nào trong danh sách có dữ liệu lương
    # → Đánh dấu -1 để sau này biết được dòng nào không được ước tính
    return -1

# BƯỚC 4: TÍNH LƯƠNG ƯỚC TÍNH CHO TOÀN BỘ BẢNG
# apply: Áp dụng hàm estimate_salary_from_list() cho từng dòng
# lambda x: Lấy cột Roles của dòng đó và tính lương ước tính
# Kết quả: Series mới chứa lương ước tính cho mỗi dòng
print("Đang tính toán mức lương ước tính...")
estimated_min_series = data_API['Roles'].apply(lambda x: estimate_salary_from_list(x, median_min_by_role))
estimated_max_series = data_API['Roles'].apply(lambda x: estimate_salary_from_list(x, median_max_by_role))

# BƯỚC 5: ĐIỀN DỮ LIỆU KHUYẾT (IMPUTATION)
# fillna(): Chỉ điền vào những ô bị thiếu (NaN)
# Những ô đã có giá trị sẽ GIỮ NGUYÊN không bị thay đổi
# VD: Salary_Min = [50M, NaN, NaN] → [50M, 47.5M, 55M] (sau điền)
print("Đang điền dữ liệu khuyết...")
data_API['Salary_Min'] = data_API['Salary_Min'].fillna(estimated_min_series)
data_API['Salary_Max'] = data_API['Salary_Max'].fillna(estimated_max_series)

# XỬ LÝ TRƯỜNG HỢP ĐẶCBIỆT
# Nếu estimate_salary_from_list() trả về -1
# Nghĩa là: Công việc có roles nhưng những roles đó không có dữ liệu lương trong bảng
# BƯỚC 6: KIỂM TRA KẾT QUẢ
# In ra số lượng lương vẫn còn thiếu sau khi điền
print(f"\nSố lượng missing Salary_Min còn lại: {data_API['Salary_Min'].isnull().sum()}")
print(f"Số lượng missing Salary_Max còn lại: {data_API['Salary_Max'].isnull().sum()}")

# BƯỚC 7: HIỂN THỊ MẪU KẾT QUẢ
# Tạo bộ lọc: Chỉ lấy những dòng mà lương được điền thành công
# (notna(): không phải NaN, != -1: không phải giá trị đánh dấu)
check_mask = (data_API['Salary_Min'].notna()) & (data_API['Salary_Min'] != -1)
print("\nXem mẫu 5 dòng dữ liệu sau khi điền:")
# Hiển thị các cột liên quan để xem kết quả điền có hợp lý không
display(data_API[check_mask][['Roles', 'Salary_Min', 'Salary_Max']].head())
```

```python
# Chuyển các giá trị missing như '', [], [UNKNOWN] thành None để phần tích missing sau đó điền dễ dàng hơn
def clean_everything_to_none(x):
    # 1. XỬ LÝ NẾU LÀ LIST
    if isinstance(x, list):
        # Nếu là list rỗng []
        if len(x) == 0:
            return None
        # Nếu là list ['UNKNOWN'] (không phân biệt hoa thường)
        if len(x) == 1 and str(x[0]).strip().upper() == 'UNKNOWN':
            return None
        return x # Trả về list nguyên vẹn nếu có dữ liệu

    # 2. XỬ LÝ NẾU LÀ STRING
    if isinstance(x, str):
        # Nếu chuỗi rỗng hoặc toàn khoảng trắng
        if x.strip() == '':
            return None
        return x

    # 3. GIỮ NGUYÊN CÁC KIỂU KHÁC (Số, ngày tháng...)
    return x

try:
    data_API = data_API.map(clean_everything_to_none)
except AttributeError:
    data_API = data_API.applymap(clean_everything_to_none)

print("Xử lý hoàn tất!")

# --- KIỂM TRA HIỆU QUẢ ---
print("\n--- Thống kê số lượng NULL (isnull) sau khi xử lý ---")
print(data_API.isnull().sum())

print("\n--- Kiểm tra mẫu cột List (Tech_Stack) ---")
# Lấy ra những dòng null để xem nó hiển thị ra sao (Nó sẽ ra None hoặc NaN)
null_sample = data_API[data_API['Tech_Stack'].isnull()]
if not null_sample.empty:
    print(f"Ví dụ dòng bị null ở Tech_Stack:\n{null_sample['Tech_Stack'].head(1)}")
else:
    print("Không còn dòng nào null ở Tech_Stack (Hoặc toàn bộ đều có dữ liệu).")
```

```python
# Tiến hành map các công ty theo key-value hiện có, với key là lĩnh vực còn values là danh sách các công ty thuộc lĩnh vực đó
company_field_map = data_API.groupby('Company_Industry')['Company_Name'].unique().to_dict()

# Nếu muốn chuyển các giá trị từ numpy array sang list:
company_field_map = {k: list(v) for k, v in company_field_map.items()}

# Xem kết quả
for field, companies in list(company_field_map.items())[:]:
    print(f"Lĩnh vực: {field}\nDanh sách công ty: {companies}\n")
```

```python
# Tiến hành điền missing values trong cột 'Company_Industry' dựa trên giá trị hiện có của chính 'Company_Name' đó, nếu rỗng thì điền 'UNKNOWN'
def fill_company_field(group):
    # Lấy giá trị lĩnh vực đầu tiên KHÁC NaN và KHÁC 'UNKNOWN'
    first_field = group[(group != 'UNKNOWN') & (pd.notnull(group))]
    if not first_field.empty:
        # Nếu tìm thấy lĩnh vực thực sự, điền cho tất cả các dòng bị thiếu hoặc 'UNKNOWN'
        return group.apply(lambda x: first_field.iloc[0] if pd.isnull(x) or x == 'UNKNOWN' else x)
    else:
        # Nếu toàn bộ group đều NaN hoặc 'UNKNOWN', điền 'UNKNOWN'
        return group.apply(lambda x: 'UNKNOWN')

data_API['Company_Industry'] = data_API.groupby('Company_Name')['Company_Industry'].transform(fill_company_field)
```

### Tạo JOB_ID cho mỗi tin đăng trước khi phân rã để đạt chuẩn 1


```python
data_1nf['JOB_ID'] = data_1nf.index + 1  # Tạo JOB_ID từ chỉ số của DataFrame, bắt đầu từ 1
# Đưa JOB_ID lên đầu tiên
```

```python
def clean_list_column(x):
    if isinstance(x, list):
        cleaned = set(item.strip() for item in x if isinstance(item, str) and item.strip() != '')
        return list(cleaned)
    return x

cols_to_explode = ['TECH_STACKS', 'PROGRAMMING_LANGUAGES', 'LOCATIONS_NAME', 'LANGUAGES_NAME', 'ROLE_NAME', 'COMPANY_LOCATION', 'SKILLS']
for col in cols_to_explode:
    data_1nf[col] = data_1nf[col].apply(clean_list_column)

data_1nf.info()
```

```python
# Explode các thuộc tính có dạng list để đạt chuẩn 1NF
for col in cols_to_explode:
    data_1nf = data_1nf.explode(col).reset_index(drop=True)
data_1nf.info()
```

```python
#Loại bỏ khoảng trắng dư thừa ở đầu và cuối chuỗi
data_1nf = data_1nf.map(lambda x: x.strip() if isinstance(x, str) else x)
```
